````python
import pandas as pd
import numpy as np

# Keep the original dataset unchanged
df_clean = df.copy()

# Remove duplicate records
df_clean = df_clean.drop_duplicates().copy()

# Remove columns that contain only one unique value
constant_cols = [
    col for col in df_clean.columns
    if df_clean[col].nunique(dropna=False) <= 1
]

df_clean.drop(columns=constant_cols, inplace=True)

print("Constant columns removed:", constant_cols)
print("Dataset shape after cleaning:", df_clean.shape)
print("Duplicate rows remaining:", df_clean.duplicated().sum())
print("Missing values remaining:", df_clean.isnull().sum().sum())

display(df_clean.head())
````

````python
# Detect likely identifier columns
id_candidates = [
    col for col in df_clean.columns
    if col.lower().replace(" ", "").replace("_", "")
    in ["employeeid", "employeenumber", "id"]
]

# Preserve attrition separately if available
target_col = next(
    (col for col in df_clean.columns
     if col.lower().replace(" ", "").replace("_", "") == "attrition"),
    None
)

employee_ids = (
    df_clean[id_candidates[0]].copy()
    if id_candidates else pd.Series(df_clean.index, index=df_clean.index)
)

attrition_labels = (
    df_clean[target_col].copy()
    if target_col else None
)

# Exclude IDs and the attrition label from clustering features
exclude_cols = id_candidates.copy()
if target_col:
    exclude_cols.append(target_col)

feature_df = df_clean.drop(columns=exclude_cols, errors="ignore").copy()

print("Identifier columns excluded:", id_candidates)
print("Attrition label preserved separately:", target_col)
print("Features available for clustering:", feature_df.shape[1])
print(feature_df.columns.tolist())
````

````python
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler, OneHotEncoder

# Identify numerical and categorical columns
numeric_cols = feature_df.select_dtypes(
    include=["number", "bool"]
).columns.tolist()

categorical_cols = feature_df.select_dtypes(
    include=["object", "category", "string"]
).columns.tolist()

# Numerical preprocessing
numeric_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

# Categorical preprocessing
categorical_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(
        handle_unknown="ignore",
        sparse_output=False
    ))
])

# Combine preprocessing steps
preprocessor = ColumnTransformer([
    ("numeric", numeric_pipeline, numeric_cols),
    ("categorical", categorical_pipeline, categorical_cols)
])

X_processed = preprocessor.fit_transform(feature_df)

# Get readable transformed feature names
feature_names = preprocessor.get_feature_names_out()

X_processed = pd.DataFrame(
    X_processed,
    columns=feature_names,
    index=feature_df.index
)

print("Preprocessing completed successfully!")
print("Processed dataset shape:", X_processed.shape)
print("Missing values:", X_processed.isnull().sum().sum())

display(X_processed.head())
````

````python
import os

os.makedirs("employee_workforce_segmentation/data/processed",
            exist_ok=True)

processed_path = (
    "employee_workforce_segmentation/data/processed/"
    "processed_employee_features.csv"
)

X_processed.to_csv(processed_path, index=False)

print("Processed data saved to:", processed_path)
print("Number of employee records:", len(X_processed))
print("Number of processed features:", X_processed.shape[1])
````

````python
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
import matplotlib.pyplot as plt

# Test different cluster counts
max_k = min(8, len(X_processed) - 1)
k_values = list(range(2, max_k + 1))

inertia_values = []
silhouette_values = []

for k in k_values:
    model = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    labels = model.fit_predict(X_processed)

    inertia_values.append(model.inertia_)
    silhouette_values.append(
        silhouette_score(X_processed, labels)
    )

# Plot Elbow Method
plt.figure(figsize=(8, 4))
plt.plot(k_values, inertia_values, marker="o")
plt.title("Elbow Method for K-Means")
plt.xlabel("Number of Clusters (K)")
plt.ylabel("Inertia")
plt.xticks(k_values)
plt.grid(True)
plt.show()

# Plot Silhouette Scores
plt.figure(figsize=(8, 4))
plt.plot(k_values, silhouette_values, marker="o")
plt.title("Silhouette Score Comparison")
plt.xlabel("Number of Clusters (K)")
plt.ylabel("Silhouette Score")
plt.xticks(k_values)
plt.grid(True)
plt.show()

best_k = k_values[silhouette_values.index(max(silhouette_values))]

print("Best K based on silhouette score:", best_k)
print("Highest silhouette score:", round(max(silhouette_values), 4))
````

````python
from sklearn.cluster import KMeans

kmeans_model = KMeans(
    n_clusters=best_k,
    random_state=42,
    n_init=10
)

kmeans_labels = kmeans_model.fit_predict(X_processed)

# Keep the cluster labels with each employee
employee_segments = df_clean.copy()
employee_segments["KMeans_Cluster"] = kmeans_labels

print("K-Means training completed!")
print("\nEmployees in each cluster:")
print(employee_segments["KMeans_Cluster"].value_counts().sort_index())

display(employee_segments.head())
````

````python
from sklearn.cluster import AgglomerativeClustering

hierarchical_model = AgglomerativeClustering(
    n_clusters=best_k,
    linkage="ward"
)

hierarchical_labels = hierarchical_model.fit_predict(X_processed)

employee_segments["Hierarchical_Cluster"] = hierarchical_labels

print("Hierarchical clustering completed!")
print("\nEmployees in each hierarchical cluster:")
print(
    employee_segments["Hierarchical_Cluster"]
    .value_counts()
    .sort_index()
)

display(employee_segments.head())
````

````python
from sklearn.metrics import (
    silhouette_score,
    davies_bouldin_score,
    calinski_harabasz_score
)
import pandas as pd

comparison = pd.DataFrame({
    "Model": ["K-Means", "Hierarchical"],
    "Silhouette Score": [
        silhouette_score(X_processed, kmeans_labels),
        silhouette_score(X_processed, hierarchical_labels)
    ],
    "Davies-Bouldin Index": [
        davies_bouldin_score(X_processed, kmeans_labels),
        davies_bouldin_score(X_processed, hierarchical_labels)
    ],
    "Calinski-Harabasz Score": [
        calinski_harabasz_score(X_processed, kmeans_labels),
        calinski_harabasz_score(X_processed, hierarchical_labels)
    ]
})

display(comparison.round(4))
````

````python
from sklearn.decomposition import PCA
import seaborn as sns
import matplotlib.pyplot as plt

pca = PCA(n_components=2)

X_pca = pca.fit_transform(X_processed)

pca_df = pd.DataFrame({
    "PC1": X_pca[:, 0],
    "PC2": X_pca[:, 1],
    "KMeans_Cluster": kmeans_labels,
    "Hierarchical_Cluster": hierarchical_labels
}, index=X_processed.index)

print("Variance explained by PC1 and PC2:")
print(pca.explained_variance_ratio_.round(4))

print(
    "Total variance explained:",
    round(pca.explained_variance_ratio_.sum() * 100, 2),
    "%"
)

fig, axes = plt.subplots(1, 2, figsize=(15, 5))

sns.scatterplot(
    data=pca_df,
    x="PC1",
    y="PC2",
    hue="KMeans_Cluster",
    palette="tab10",
    s=35,
    ax=axes[0]
)
axes[0].set_title("PCA Visualization — K-Means")

sns.scatterplot(
    data=pca_df,
    x="PC1",
    y="PC2",
    hue="Hierarchical_Cluster",
    palette="tab10",
    s=35,
    ax=axes[1]
)
axes[1].set_title("PCA Visualization — Hierarchical")

plt.tight_layout()
plt.show()
````

````python
# Cluster sizes
print("Cluster sizes:")
display(
    employee_segments["KMeans_Cluster"]
    .value_counts()
    .sort_index()
    .rename("Employee Count")
    .to_frame()
)

# Select useful columns that exist in the dataset
profile_candidates = [
    "Age",
    "MonthlyIncome",
    "MonthlyRate",
    "JobSatisfaction",
    "EnvironmentSatisfaction",
    "RelationshipSatisfaction",
    "WorkLifeBalance",
    "PerformanceRating",
    "YearsAtCompany",
    "YearsInCurrentRole",
    "JobLevel",
    "DistanceFromHome"
]

numeric_profile_cols = [
    col for col in profile_candidates
    if col in employee_segments.columns
]

if numeric_profile_cols:
    print("Average employee characteristics by cluster:")
    display(
        employee_segments.groupby("KMeans_Cluster")[
            numeric_profile_cols
        ].mean().round(2)
    )

if "Department" in employee_segments.columns:
    print("Department distribution within each cluster:")
    display(
        pd.crosstab(
            employee_segments["KMeans_Cluster"],
            employee_segments["Department"],
            normalize="index"
        ).round(3) * 100
    )

if "Attrition" in employee_segments.columns:
    print("Historical attrition rate by cluster:")
    attrition_rate = (
        employee_segments.assign(
            Attrition_Flag=(
                employee_segments["Attrition"]
                .astype(str).str.strip().str.lower()
                .map({"yes": 1, "no": 0})
            )
        )
        .groupby("KMeans_Cluster")["Attrition_Flag"]
        .mean()
        .mul(100)
        .round(2)
    )

    display(attrition_rate.rename(
        "Historical Attrition (%)"
    ).to_frame())
````

````python
# Analyze available employee performance and satisfaction fields
analysis_cols = [
    "PerformanceRating",
    "JobSatisfaction",
    "EnvironmentSatisfaction",
    "RelationshipSatisfaction",
    "WorkLifeBalance",
    "JobLevel",
    "YearsAtCompany",
    "MonthlyIncome"
]

available_cols = [
    col for col in analysis_cols
    if col in employee_segments.columns
]

if available_cols:
    cluster_profile = (
        employee_segments
        .groupby("KMeans_Cluster")[available_cols]
        .mean()
        .round(2)
    )

    print("Employee performance and satisfaction by cluster:")
    display(cluster_profile)

    # Visualize fields that exist in the dataset
    for col in [
        "PerformanceRating",
        "JobSatisfaction",
        "WorkLifeBalance"
    ]:
        if col in employee_segments.columns:
            plt.figure(figsize=(8, 4))
            sns.boxplot(
                data=employee_segments,
                x="KMeans_Cluster",
                y=col
            )
            plt.title(f"{col} Across Employee Clusters")
            plt.tight_layout()
            plt.show()
else:
    print("No matching performance or satisfaction columns found.")
````

````python
if "Attrition" in employee_segments.columns:

    attrition_text = (
        employee_segments["Attrition"]
        .astype("string")
        .str.strip()
        .str.lower()
    )

    # Handle common yes/no and true/false representations
    attrition_flag = attrition_text.map({
        "yes": 1,
        "no": 0,
        "true": 1,
        "false": 0,
        "1": 1,
        "0": 0
    })

    retention_df = employee_segments.copy()
    retention_df["Attrition_Flag"] = attrition_flag

    retention_summary = (
        retention_df
        .groupby("KMeans_Cluster")
        .agg(
            Employee_Count=("Attrition_Flag", "size"),
            Known_Attrition_Records=("Attrition_Flag", "count"),
            Historical_Attrition_Rate=("Attrition_Flag", "mean")
        )
    )

    retention_summary["Historical_Attrition_Percent"] = (
        retention_summary["Historical_Attrition_Rate"] * 100
    ).round(2)

    overall_rate = attrition_flag.mean() * 100

    print(f"Overall historical attrition rate: {overall_rate:.2f}%")
    print("Historical attrition by cluster:")
    display(retention_summary.round(3))

    plt.figure(figsize=(8, 4))
    retention_summary["Historical_Attrition_Percent"].plot(
        kind="bar"
    )
    plt.axhline(
        overall_rate,
        linestyle="--",
        label="Overall dataset rate"
    )
    plt.title("Historical Attrition Rate by Employee Cluster")
    plt.xlabel("K-Means Cluster")
    plt.ylabel("Historical Attrition (%)")
    plt.legend()
    plt.tight_layout()
    plt.show()

else:
    print(
        "No Attrition column found. "
        "Historical retention analysis requires an attrition label."
    )
````

````python

%%writefile app.py
import os
import pandas as pd
import streamlit as st
import plotly.express as px

st.set_page_config(
    page_title="Employee Workforce Segmentation",
    page_icon="📊",
    layout="wide"
)

RESULTS_DIR = "employee_workforce_segmentation/results"

st.title("Employee Workforce Segmentation")
st.caption(
    "Workforce analytics | Clustering | PCA | Department insights | Retention"
)

@st.cache_data
def load_csv(path):
    return pd.read_csv(path)

results_path = os.path.join(
    RESULTS_DIR, "employee_segmentation_results.csv"
)
pca_path = os.path.join(
    RESULTS_DIR, "pca_employee_visualization.csv"
)
comparison_path = os.path.join(
    RESULTS_DIR, "clustering_model_comparison.csv"
)

if not os.path.exists(results_path):
    st.error(
        "Employee segmentation results were not found. "
        "Place employee_segmentation_results.csv inside "
        "employee_workforce_segmentation/results/."
    )
    st.stop()

df = load_csv(results_path)

if df.empty:
    st.error("The employee segmentation file is empty.")
    st.stop()

# Sidebar filters
st.sidebar.header("Dashboard Filters")

if "Department" in df.columns:
    departments = sorted(
        df["Department"].dropna().astype(str).unique().tolist()
    )
    selected_departments = st.sidebar.multiselect(
        "Select department",
        departments,
        default=departments
    )
    filtered = df[
        df["Department"].astype(str).isin(selected_departments)
    ].copy()
else:
    filtered = df.copy()
    st.sidebar.info("Department field is not available.")

cluster_options = []
if "KMeans_Cluster" in filtered.columns:
    cluster_options = sorted(
        filtered["KMeans_Cluster"].dropna().unique().tolist()
    )

if cluster_options:
    selected_clusters = st.sidebar.multiselect(
        "Select K-Means clusters",
        cluster_options,
        default=cluster_options
    )
    filtered = filtered[
        filtered["KMeans_Cluster"].isin(selected_clusters)
    ].copy()

if filtered.empty:
    st.warning("No employees match the selected filters.")
    st.stop()

# KPI metrics
st.header("Workforce Overview")

c1, c2, c3, c4 = st.columns(4)
c1.metric("Employees", f"{len(filtered):,}")

if "Department" in filtered.columns:
    c2.metric("Departments", filtered["Department"].nunique())
else:
    c2.metric("Departments", "N/A")

if "KMeans_Cluster" in filtered.columns:
    c3.metric("K-Means Clusters", filtered["KMeans_Cluster"].nunique())
else:
    c3.metric("K-Means Clusters", "N/A")

if "MonthlyIncome" in filtered.columns:
    c4.metric(
        "Average Monthly Income",
        f"{filtered['MonthlyIncome'].mean():,.0f}"
    )
else:
    c4.metric("Average Monthly Income", "N/A")

# Cluster distribution
st.header("Employee Segmentation")

left, right = st.columns(2)

with left:
    if "KMeans_Cluster" in filtered.columns:
        cluster_counts = (
            filtered["KMeans_Cluster"]
            .value_counts()
            .sort_index()
            .rename_axis("Cluster")
            .reset_index(name="Employees")
        )
        fig = px.bar(
            cluster_counts,
            x="Cluster",
            y="Employees",
            title="K-Means Cluster Distribution",
            text="Employees"
        )
        st.plotly_chart(fig, use_container_width=True)

with right:
    if "Hierarchical_Cluster" in filtered.columns:
        hierarchy_counts = (
            filtered["Hierarchical_Cluster"]
            .value_counts()
            .sort_index()
            .rename_axis("Cluster")
            .reset_index(name="Employees")
        )
        fig = px.bar(
            hierarchy_counts,
            x="Cluster",
            y="Employees",
            title="Hierarchical Cluster Distribution",
            text="Employees"
        )
        st.plotly_chart(fig, use_container_width=True)

# Department segmentation
if "Department" in filtered.columns and "KMeans_Cluster" in filtered.columns:
    st.header("Department-Level Workforce Segments")

    department_counts = pd.crosstab(
        filtered["Department"],
        filtered["KMeans_Cluster"]
    ).reset_index()

    department_long = department_counts.melt(
        id_vars="Department",
        var_name="Cluster",
        value_name="Employees"
    )

    fig = px.bar(
        department_long,
        x="Department",
        y="Employees",
        color="Cluster",
        barmode="stack",
        title="Department Distribution by K-Means Cluster"
    )
    st.plotly_chart(fig, use_container_width=True)

# PCA visualization
st.header("PCA Cluster Visualization")

if os.path.exists(pca_path):
    pca_df = load_csv(pca_path)

    if {"PC1", "PC2"}.issubset(pca_df.columns):
        label_col = "KMeans_Cluster"
        if label_col not in pca_df.columns:
            label_col = "Hierarchical_Cluster"

        if label_col in pca_df.columns:
            # Align PCA rows to the employee table by original row order.
            # This works when both exports retain the same row order.
            if len(pca_df) == len(df):
                selected_positions = df.index[
                    df.index.isin(filtered.index)
                ]
                pca_filtered = pca_df.iloc[
                    selected_positions.tolist()
                ].copy()
            else:
                pca_filtered = pca_df.copy()

            fig = px.scatter(
                pca_filtered,
                x="PC1",
                y="PC2",
                color=label_col,
                title="Employees Projected onto Two Principal Components",
                opacity=0.75
            )
            st.plotly_chart(fig, use_container_width=True)
            st.caption(
                "PCA is a two-dimensional projection of the processed "
                "features. Overlapping points do not necessarily indicate "
                "identical employees."
            )
        else:
            st.info("Cluster labels were not found in the PCA file.")
    else:
        st.info("The PCA file must contain PC1 and PC2 columns.")
else:
    st.info("PCA results file is not available.")

# Performance and satisfaction
profile_cols = [
    "PerformanceRating",
    "JobSatisfaction",
    "EnvironmentSatisfaction",
    "WorkLifeBalance",
    "YearsAtCompany",
    "MonthlyIncome"
]
available_profile_cols = [
    col for col in profile_cols if col in filtered.columns
]

if available_profile_cols and "KMeans_Cluster" in filtered.columns:
    st.header("Performance and Employee Characteristics")
    profile = (
        filtered.groupby("KMeans_Cluster")[available_profile_cols]
        .mean()
        .round(2)
    )
    st.dataframe(profile, use_container_width=True)

# Historical attrition
if "Attrition" in filtered.columns:
    st.header("Historical Attrition Analysis")

    attrition = (
        filtered["Attrition"]
        .astype(str)
        .str.strip()
        .str.lower()
        .map({
            "yes": 1, "no": 0,
            "true": 1, "false": 0,
            "1": 1, "0": 0
        })
    )

    known = filtered.assign(Attrition_Flag=attrition).dropna(
        subset=["Attrition_Flag"]
    )

    if not known.empty:
        overall_rate = known["Attrition_Flag"].mean() * 100
        st.metric("Historical Attrition Rate", f"{overall_rate:.2f}%")

        if "KMeans_Cluster" in known.columns:
            rates = (
                known.groupby("KMeans_Cluster")["Attrition_Flag"]
                .mean()
                .mul(100)
                .reset_index(name="Historical Attrition (%)")
            )
            fig = px.bar(
                rates,
                x="KMeans_Cluster",
                y="Historical Attrition (%)",
                title="Historical Attrition by Cluster"
            )
            st.plotly_chart(fig, use_container_width=True)

        st.caption(
            "These are historical group-level statistics, not predictions "
            "of individual employees' future behavior."
        )
    else:
        st.info("No recognizable attrition labels were found.")

# Model comparison
if os.path.exists(comparison_path):
    st.header("Clustering Model Comparison")
    comparison = load_csv(comparison_path)
    st.dataframe(comparison, use_container_width=True)

# Employee data and download
st.header("Employee Segmentation Results")
st.dataframe(filtered, use_container_width=True)

st.download_button(
    "Download Filtered Employee Data (CSV)",
    data=filtered.to_csv(index=False).encode("utf-8"),
    file_name="employee_segmentation_results.csv",
    mime="text/csv"
)

st.success("Dashboard data loaded successfully.")
````

````python
import os
import shutil

target = "/content/employee_workforce_segmentation/results"
os.makedirs(target, exist_ok=True)

required = [
    "employee_segmentation_results.csv",
    "pca_employee_visualization.csv",
    "clustering_model_comparison.csv"
]

for filename in required:
    matches = []

    for root, dirs, files in os.walk("/content"):
        dirs[:] = [
            d for d in dirs
            if d not in [".config", ".cache"]
        ]

        if filename in files:
            source = os.path.join(root, filename)
            if os.path.abspath(source) != os.path.abspath(
                os.path.join(target, filename)
            ):
                matches.append(source)

    destination = os.path.join(target, filename)

    if os.path.exists(destination):
        print("OK:", filename)
    elif matches:
        shutil.copy2(matches[0], destination)
        print("Restored:", filename, "from", matches[0])
    else:
        print("MISSING:", filename)

print("\nFinal results folder:")
print(os.listdir(target))
````

````python
import os
import pandas as pd
import streamlit as st
import plotly.express as px

st.set_page_config(
    page_title="Employee Workforce Segmentation",
    page_icon="📊",
    layout="wide"
)

RESULTS_DIR = "employee_workforce_segmentation/results"

st.title("Employee Workforce Segmentation")
st.caption(
    "Workforce analytics | Clustering | PCA | Department insights | Retention"
)

@st.cache_data
def load_csv(path):
    return pd.read_csv(path)

results_path = os.path.join(
    RESULTS_DIR, "employee_segmentation_results.csv"
)
pca_path = os.path.join(
    RESULTS_DIR, "pca_employee_visualization.csv"
)
comparison_path = os.path.join(
    RESULTS_DIR, "clustering_model_comparison.csv"
)

if not os.path.exists(results_path):
    st.error(
        "Employee segmentation results were not found. "
        "Place employee_segmentation_results.csv inside "
        "employee_workforce_segmentation/results/."
    )
    st.stop()

df = load_csv(results_path)

if df.empty:
    st.error("The employee segmentation file is empty.")
    st.stop()

# Sidebar filters
st.sidebar.header("Dashboard Filters")

if "Department" in df.columns:
    departments = sorted(
        df["Department"].dropna().astype(str).unique().tolist()
    )
    selected_departments = st.sidebar.multiselect(
        "Select department",
        departments,
        default=departments
    )
    filtered = df[
        df["Department"].astype(str).isin(selected_departments)
    ].copy()
else:
    filtered = df.copy()
    st.sidebar.info("Department field is not available.")

cluster_options = []
if "KMeans_Cluster" in filtered.columns:
    cluster_options = sorted(
        filtered["KMeans_Cluster"].dropna().unique().tolist()
    )

if cluster_options:
    selected_clusters = st.sidebar.multiselect(
        "Select K-Means clusters",
        cluster_options,
        default=cluster_options
    )
    filtered = filtered[
        filtered["KMeans_Cluster"].isin(selected_clusters)
    ].copy()

if filtered.empty:
    st.warning("No employees match the selected filters.")
    st.stop()

# KPI metrics
st.header("Workforce Overview")

c1, c2, c3, c4 = st.columns(4)
c1.metric("Employees", f"{len(filtered):,}")

if "Department" in filtered.columns:
    c2.metric("Departments", filtered["Department"].nunique())
else:
    c2.metric("Departments", "N/A")

if "KMeans_Cluster" in filtered.columns:
    c3.metric("K-Means Clusters", filtered["KMeans_Cluster"].nunique())
else:
    c3.metric("K-Means Clusters", "N/A")

if "MonthlyIncome" in filtered.columns:
    c4.metric(
        "Average Monthly Income",
        f"{filtered['MonthlyIncome'].mean():,.0f}"
    )
else:
    c4.metric("Average Monthly Income", "N/A")

# Cluster distribution
st.header("Employee Segmentation")

left, right = st.columns(2)

with left:
    if "KMeans_Cluster" in filtered.columns:
        cluster_counts = (
            filtered["KMeans_Cluster"]
            .value_counts()
            .sort_index()
            .rename_axis("Cluster")
            .reset_index(name="Employees")
        )
        fig = px.bar(
            cluster_counts,
            x="Cluster",
            y="Employees",
            title="K-Means Cluster Distribution",
            text="Employees"
        )
        st.plotly_chart(fig, use_container_width=True)

with right:
    if "Hierarchical_Cluster" in filtered.columns:
        hierarchy_counts = (
            filtered["Hierarchical_Cluster"]
            .value_counts()
            .sort_index()
            .rename_axis("Cluster")
            .reset_index(name="Employees")
        )
        fig = px.bar(
            hierarchy_counts,
            x="Cluster",
            y="Employees",
            title="Hierarchical Cluster Distribution",
            text="Employees"
        )
        st.plotly_chart(fig, use_container_width=True)

# Department segmentation
if "Department" in filtered.columns and "KMeans_Cluster" in filtered.columns:
    st.header("Department-Level Workforce Segments")

    department_counts = pd.crosstab(
        filtered["Department"],
        filtered["KMeans_Cluster"]
    ).reset_index()

    department_long = department_counts.melt(
        id_vars="Department",
        var_name="Cluster",
        value_name="Employees"
    )

    fig = px.bar(
        department_long,
        x="Department",
        y="Employees",
        color="Cluster",
        barmode="stack",
        title="Department Distribution by K-Means Cluster"
    )
    st.plotly_chart(fig, use_container_width=True)

# PCA visualization
st.header("PCA Cluster Visualization")

if os.path.exists(pca_path):
    pca_df = load_csv(pca_path)

    if {"PC1", "PC2"}.issubset(pca_df.columns):
        label_col = "KMeans_Cluster"
        if label_col not in pca_df.columns:
            label_col = "Hierarchical_Cluster"

        if label_col in pca_df.columns:
            # Align PCA rows to the employee table by original row order.
            # This works when both exports retain the same row order.
            if len(pca_df) == len(df):
                selected_positions = df.index[
                    df.index.isin(filtered.index)
                ]
                pca_filtered = pca_df.iloc[
                    selected_positions.tolist()
                ].copy()
            else:
                pca_filtered = pca_df.copy()

            fig = px.scatter(
                pca_filtered,
                x="PC1",
                y="PC2",
                color=label_col,
                title="Employees Projected onto Two Principal Components",
                opacity=0.75
            )
            st.plotly_chart(fig, use_container_width=True)
            st.caption(
                "PCA is a two-dimensional projection of the processed "
                "features. Overlapping points do not necessarily indicate "
                "identical employees."
            )
        else:
            st.info("Cluster labels were not found in the PCA file.")
    else:
        st.info("The PCA file must contain PC1 and PC2 columns.")
else:
    st.info("PCA results file is not available.")

# Performance and satisfaction
profile_cols = [
    "PerformanceRating",
    "JobSatisfaction",
    "EnvironmentSatisfaction",
    "WorkLifeBalance",
    "YearsAtCompany",
    "MonthlyIncome"
]
available_profile_cols = [
    col for col in profile_cols if col in filtered.columns
]

if available_profile_cols and "KMeans_Cluster" in filtered.columns:
    st.header("Performance and Employee Characteristics")
    profile = (
        filtered.groupby("KMeans_Cluster")[available_profile_cols]
        .mean()
        .round(2)
    )
    st.dataframe(profile, use_container_width=True)

# Historical attrition
if "Attrition" in filtered.columns:
    st.header("Historical Attrition Analysis")

    attrition = (
        filtered["Attrition"]
        .astype(str)
        .str.strip()
        .str.lower()
        .map({
            "yes": 1, "no": 0,
            "true": 1, "false": 0,
            "1": 1, "0": 0
        })
    )

    known = filtered.assign(Attrition_Flag=attrition).dropna(
        subset=["Attrition_Flag"]
    )

    if not known.empty:
        overall_rate = known["Attrition_Flag"].mean() * 100
        st.metric("Historical Attrition Rate", f"{overall_rate:.2f}%")

        if "KMeans_Cluster" in known.columns:
            rates = (
                known.groupby("KMeans_Cluster")["Attrition_Flag"]
                .mean()
                .mul(100)
                .reset_index(name="Historical Attrition (%)")
            )
            fig = px.bar(
                rates,
                x="KMeans_Cluster",
                y="Historical Attrition (%)",
                title="Historical Attrition by Cluster"
            )
            st.plotly_chart(fig, use_container_width=True)

        st.caption(
            "These are historical group-level statistics, not predictions "
            "of individual employees' future behavior."
        )
    else:
        st.info("No recognizable attrition labels were found.")

# Model comparison
if os.path.exists(comparison_path):
    st.header("Clustering Model Comparison")
    comparison = load_csv(comparison_path)
    st.dataframe(comparison, use_container_width=True)

# Employee data and download
st.header("Employee Segmentation Results")
st.dataframe(filtered, use_container_width=True)

st.download_button(
    "Download Filtered Employee Data (CSV)",
    data=filtered.to_csv(index=False).encode("utf-8"),
    file_name="employee_segmentation_results.csv",
    mime="text/csv"
)

st.success("Dashboard data loaded successfully.")

````
